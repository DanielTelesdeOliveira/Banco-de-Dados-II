###Consulta 1 — Extração com ->, ->> e #>>
O operador -> retorna o valor como JSONB (preserva o tipo original). O ->> retorna como TEXT, útil para comparações e exibição. O #>> navega por um caminho aninhado definido como array de chaves e retorna TEXT. Aqui extraímos a marca em dois formatos e o campo ram que está dois níveis abaixo, dentro de especificacoes.



```python
%%sql
SELECT
    nome,
    detalhes -> 'marca'                  AS marca_jsonb,
    detalhes ->> 'marca'                 AS marca_texto,
    detalhes ->> 'preco'                 AS preco,
    detalhes #>> '{especificacoes, ram}' AS ram
FROM produtos
ORDER BY nome;
```

###Consulta 2 — Caminho duplamente aninhado com #>>
O operador #>> aceita caminhos de qualquer profundidade. Aqui navegamos três níveis: especificacoes → configuracoes → temperatura → min/max. Isso é útil para extrair valores dentro de objetos dentro de objetos, sem precisar encadear múltiplos operadores.

```python
%%sql
SELECT
    nome,
    detalhes #>> '{especificacoes, configuracoes, temperatura, min}' AS temp_min,
    detalhes #>> '{especificacoes, configuracoes, temperatura, max}' AS temp_max
FROM produtos
WHERE nome = 'Cafeteira';
```

###Consulta 3 — Contenção @> (o JSONB contém o valor?)
O operador @> verifica se o JSONB da coluna contém o documento fornecido à direita. Não exige igualdade total, apenas que os campos especificados existam com os valores indicados. Aqui buscamos todos os produtos cuja marca é AMD.

```python
%%sql
SELECT nome, detalhes ->> 'marca' AS marca
FROM produtos
WHERE detalhes @> '{"marca": "AMD"}';
```

###Consulta 4 — Contenção @> em objeto aninhado
O @> também funciona com objetos aninhados. O PostgreSQL verifica recursivamente se a estrutura fornecida está contida no JSONB. Aqui encontramos produtos que possuem monitor_cardiaco: true dentro de especificacoes.funcoes, independente dos outros campos presentes.

```python
%%sql
SELECT nome
FROM produtos
WHERE detalhes @> '{"especificacoes": {"funcoes": {"monitor_cardiaco": true}}}';
```

###Consulta 5 — Contenção <@ (está contido no JSONB?)
O operador <@ é o inverso do @>: verifica se o documento da esquerda está contido no JSONB da coluna. Útil para validar se um registro possui ao menos um conjunto mínimo de campos com valores específicos. Aqui confirmamos que o iPhone tem exatamente a marca Apple e o preço 7599.99.

```python
%%sql
SELECT nome
FROM produtos
WHERE '{"marca": "Apple", "preco": 7599.99}'::jsonb <@ detalhes;
```

###Consulta 6 — Existência de chave com ?
O operador ? verifica se uma chave existe no nível raiz do JSONB, sem importar o valor. Diferente do @>, não compara valor — só presença. Retorna todos os produtos que possuem o campo cor declarado.

```python
%%sql
SELECT nome, detalhes ->> 'cor' AS cor
FROM produtos
WHERE detalhes ? 'cor';
```

###Consulta 7 — Existência de ao menos uma chave com ?|
O operador ?| retorna true se ao menos uma das chaves do array existir no JSONB. É equivalente a um OR de múltiplos ?. Aqui buscamos produtos que tenham cor ou modelo — ou ambos — no nível raiz.

```python
%%sql
SELECT nome
FROM produtos
WHERE detalhes ?| ARRAY['cor', 'modelo'];
```

###Consulta 8 — Existência de todas as chaves com ?&
O operador ?& retorna true somente se todas as chaves do array existirem no JSONB. É equivalente a um AND de múltiplos ?. Aqui filtramos produtos que possuem simultaneamente avaliacoes e itens_relacionados.

```python
%%sql
SELECT nome
FROM produtos
WHERE detalhes ?& ARRAY['avaliacoes', 'itens_relacionados'];
```

###Consulta 9 — JSONPath: jsonb_path_exists
jsonb_path_exists verifica se um caminho JSONPath retorna ao menos um resultado. A sintaxe $.avaliacoes[*] ? (@.nota == 5) percorre todos os elementos do array avaliacoes e filtra os que têm nota igual a 5. Retorna os produtos que receberam nota máxima de ao menos um usuário.

```python
%%sql
SELECT nome
FROM produtos
WHERE jsonb_path_exists(detalhes, '$.avaliacoes[*] ? (@.nota == 5)');
```

###Consulta 10 — JSONPath: jsonb_path_query extraindo campos
jsonb_path_query retorna todos os valores que o caminho JSONPath encontrar, gerando uma linha por resultado. Aqui extraímos apenas o campo comentario das avaliações com nota maior ou igual a 4, para todos os produtos.

```python
%%sql
SELECT
    nome,
    jsonb_path_query(detalhes, '$.avaliacoes[*] ? (@.nota >= 4).comentario') AS comentario
FROM produtos
ORDER BY nome;
```

###Consulta 11 — JSONPath: filtro numérico no campo preco
JSONPath suporta comparações numéricas diretamente. Combinamos jsonb_path_exists no WHERE para filtrar produtos com preço acima de 500, e jsonb_path_query no SELECT para exibir o valor. Isso demonstra que JSONPath pode substituir conversões manuais de ->> para NUMERIC em filtros simples.

```python
%%sql
SELECT
    nome,
    jsonb_path_query(detalhes, '$.preco') AS preco
FROM produtos
WHERE jsonb_path_exists(detalhes, '$.preco ? (@ > 500)');
```

###Consulta 12 — jsonb_set: atualiza um campo existente
jsonb_set(target, path, new_value) substitui o valor de um campo dentro do JSONB sem reescrever o documento inteiro. O caminho {preco} aponta para a chave raiz preco. Após o UPDATE, fazemos um SELECT para confirmar a alteração.

```python
%%sql
UPDATE produtos
SET
    detalhes   = jsonb_set(detalhes, '{preco}', '480.00'),
    updated_at = NOW()
WHERE nome = 'Drone';

SELECT nome, detalhes ->> 'preco' AS preco_atualizado
FROM produtos
WHERE nome = 'Drone';
```

###Consulta 13 — jsonb_set: adiciona um campo novo
Quando o quarto parâmetro de jsonb_set é true, o campo é criado caso não exista. Aqui adicionamos em_estoque: true ao Notebook sem alterar nenhum outro campo do JSONB. Isso demonstra como enriquecer registros com novos atributos sem precisar de ALTER TABLE.

```python
%%sql
UPDATE produtos
SET
    detalhes   = jsonb_set(detalhes, '{em_estoque}', 'true', true),
    updated_at = NOW()
WHERE nome = 'Notebook';

SELECT nome, detalhes -> 'em_estoque' AS em_estoque
FROM produtos
WHERE nome = 'Notebook';
```

###Consulta 14 — jsonb_insert: insere item em array
jsonb_insert(target, path, new_value, insert_after) insere um elemento em um array JSONB sem sobrescrever os existentes. O caminho {avaliacoes, -1} aponta para após o último elemento. Com insert_after = true, o novo objeto é inserido ao final da lista de avaliações do Tablet.

```python
%%sql
UPDATE produtos
SET
    detalhes   = jsonb_insert(
                     detalhes,
                     '{avaliacoes, -1}',
                     '{"usuario": "Camila", "nota": 5, "comentario": "Excelente para ensino a distância"}',
                     true
                 ),
    updated_at = NOW()
WHERE nome = 'Tablet';

SELECT nome, detalhes -> 'avaliacoes' AS avaliacoes
FROM produtos
WHERE nome = 'Tablet';
```

###Consulta 15 — jsonb_array_elements: média de notas por produto
jsonb_array_elements expande cada elemento de um array JSONB em uma linha separada, funcionando como um CROSS JOIN. Aqui expandimos o array avaliacoes de cada produto, convertemos nota para NUMERIC e calculamos a média agrupando por produto.

```python
%%sql
SELECT
    p.nome,
    COUNT(*)                                          AS total_avaliacoes,
    ROUND(AVG((av.value ->> 'nota')::NUMERIC), 2)    AS media_nota
FROM produtos p,
     jsonb_array_elements(p.detalhes -> 'avaliacoes') AS av
GROUP BY p.nome
ORDER BY media_nota DESC;
```

###Consulta 16 — jsonb_array_elements: lista de usuários e notas
Reutilizamos jsonb_array_elements para extrair individualmente o usuario e a nota de cada avaliação de cada produto. O ->> converte o campo para TEXT e o cast ::INT para inteiro. O DISTINCT evita duplicatas caso o mesmo usuário apareça em produtos diferentes com mesma nota.

```python
%%sql
SELECT DISTINCT
    p.nome                     AS produto,
    av.value ->> 'usuario'     AS usuario,
    (av.value ->> 'nota')::INT AS nota
FROM produtos p,
     jsonb_array_elements(p.detalhes -> 'avaliacoes') AS av
ORDER BY nota DESC, produto;
```

###Consulta 17 — jsonb_each: expande chaves de um objeto
jsonb_each transforma as chaves de um objeto JSONB em linhas com colunas key e value. Diferente de jsonb_array_elements que trabalha com arrays, jsonb_each trabalha com objetos. Aqui expandimos todas as chaves de especificacoes do Smartwatch, exibindo cada propriedade em uma linha separada.

```python
%%sql
SELECT
    key   AS campo,
    value AS valor
FROM produtos,
     jsonb_each(detalhes -> 'especificacoes')
WHERE nome = 'Smartwatch';
```

###Consulta 18 — Criação do índice GIN
O índice GIN (Generalized Inverted Index) é o tipo recomendado para colunas JSONB. Ele indexa cada chave e valor individualmente, acelerando operadores como @>, <@, ?, ?| e ?&. Sem ele, essas buscas fazem varredura sequencial em toda a tabela.

```python
%%sql
CREATE INDEX IF NOT EXISTS idx_detalhes_gin
ON produtos
USING GIN (detalhes);
```

###Consulta 19 — EXPLAIN ANALYZE com @>
EXPLAIN ANALYZE executa a consulta e exibe o plano de execução com tempos reais. Após criar o índice GIN, esperamos ver Bitmap Index Scan no plano, indicando que o índice foi usado. Sem o índice, apareceria Seq Scan, que percorre todos os registros um a um.

```python
%%sql
EXPLAIN ANALYZE
SELECT nome
FROM produtos
WHERE detalhes @> '{"marca": "AMD"}';
```

###Consulta 20 — EXPLAIN ANALYZE com ?
Validamos o uso do índice GIN também com o operador ? de existência de chave. A busca por em_estoque demonstra que o índice GIN cobre tanto buscas por valor (@>) quanto por presença de chave (?), tornando ambas as operações eficientes mesmo em tabelas grandes.

```python
%%sql
EXPLAIN ANALYZE
SELECT nome
FROM produtos
WHERE detalhes ? 'em_estoque';
```
